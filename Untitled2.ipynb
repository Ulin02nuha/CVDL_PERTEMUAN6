```python
import os
import sys
import time
import copy
import shutil
import glob
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from PIL import Image

import torch
import torch.nn as nn
import torch.optim as optim
from torch.utils.data import DataLoader
from torchvision import datasets, transforms
import timm
from sklearn.metrics import classification_report, confusion_matrix

# Konfigurasi Seed & Device GPU
def set_seed(seed=42):
    torch.manual_seed(seed)
    torch.cuda.manual_seed_all(seed)
    np.random.seed(seed)
    torch.backends.cudnn.deterministic = True

set_seed(42)
device = torch.device("cuda:0" if torch.cuda.is_available() else "cpu")
print(f"[INFO] Device yang digunakan: {device}")
```

```python
DATA_DIR = './data_fruit360_filtered'
TRAIN_DIR = os.path.join(DATA_DIR, 'train')
TEST_DIR = os.path.join(DATA_DIR, 'test')

# Pemetaan sub-folder Fruits-360 ke 3 kelas utama
CLASS_MAPPING = {
    'Apple': ['Apple Braeburn', 'Apple Crimson Snow', 'Apple Golden 1', 'Apple Golden 2',
              'Apple Golden 3', 'Apple Granny Smith', 'Apple Red 1', 'Apple Red 2',
              'Apple Red 3', 'Apple Red Delicious', 'Apple Red Yellow 1', 'Apple Red Yellow 2'],
    'Cherry': ['Cherry 1', 'Cherry 2', 'Cherry Rainier', 'Cherry Wax Red',
               'Cherry Wax Calypso', 'Cherry Wax Yellow'],
    'Tomato': ['Tomato 1', 'Tomato 2', 'Tomato 3', 'Tomato 4',
               'Tomato Cherry Red', 'Tomato Maroon', 'Tomato Yellow']
}

def prepare_fruits360_dataset():
    if os.path.exists(TRAIN_DIR) and len(os.listdir(TRAIN_DIR)) == 3:
        print("[INFO] Dataset 3 kelas terdeteksi dan siap digunakan.")
        return

    print("[INFO] Mengunduh dataset Fruits-360 dari Kaggle...")
    os.environ['KAGGLE_CONFIG_DIR'] = '/content'

    !kaggle datasets download -d gaborfodor/fruits-360-for-uploaded-images --unzip -p /content/fruits360_raw > /dev/null 2>&1

    if not os.path.exists('/content/fruits360_raw'):
        !kaggle datasets download -d meharunnessa/fruits-360 --unzip -p /content/fruits360_raw > /dev/null 2>&1

    print("[INFO] Menggabungkan folder ke 3 kelas utama (Apple, Cherry, Tomato)...")
    raw_root = '/content/fruits360_raw'
    train_raw = glob.glob(os.path.join(raw_root, '**', 'Training'), recursive=True)
    test_raw = glob.glob(os.path.join(raw_root, '**', 'Test'), recursive=True)

    if not train_raw or not test_raw:
        print("[WARNING] Kredensial Kaggle tidak ada. Membuat sampel otomatis untuk tes kode...")
        for split, target_dir in [('train', TRAIN_DIR), ('test', TEST_DIR)]:
            for target_class in CLASS_MAPPING.keys():
                c_dir = os.path.join(target_dir, target_class)
                os.makedirs(c_dir, exist_ok=True)
                count = 30 if split == 'train' else 15
                for i in range(count):
                    img = Image.fromarray(np.random.randint(0, 256, (300, 300, 3), dtype=np.uint8))
                    img.save(os.path.join(c_dir, f'sample_{i}.jpg'))
        return

    for split_name, src_dir, dst_dir in [('Train', train_raw[0], TRAIN_DIR), ('Test', test_raw[0], TEST_DIR)]:
        for target_class, subfolders in CLASS_MAPPING.items():
            target_path = os.path.join(dst_dir, target_class)
            os.makedirs(target_path, exist_ok=True)
            for sub in subfolders:
                src_sub_path = os.path.join(src_dir, sub)
                if os.path.exists(src_sub_path):
                    for file_name in os.listdir(src_sub_path):
                        src_file = os.path.join(src_sub_path, file_name)
                        dst_file = os.path.join(target_path, f"{sub}_{file_name}")
                        shutil.copy(src_file, dst_file)

    print("[INFO] Penyiapan dataset selesai.")

prepare_fruits360_dataset()
```

```python
# Transformasi Preprocessing Inception-v4: Resize 341 -> CenterCrop 299x299 -> Normalisasi
data_transforms = {
    'train': transforms.Compose([
        transforms.Resize(341),
        transforms.CenterCrop(299),
        transforms.ToTensor(),
        transforms.Normalize(mean=[0.5, 0.5, 0.5], std=[0.5, 0.5, 0.5])
    ]),
    'test': transforms.Compose([
        transforms.Resize(341),
        transforms.CenterCrop(299),
        transforms.ToTensor(),
        transforms.Normalize(mean=[0.5, 0.5, 0.5], std=[0.5, 0.5, 0.5])
    ]),
}

image_datasets = {
    'train': datasets.ImageFolder(TRAIN_DIR, transform=data_transforms['train']),
    'test': datasets.ImageFolder(TEST_DIR, transform=data_transforms['test'])
}

BATCH_SIZE = 16

dataloaders = {
    'train': DataLoader(image_datasets['train'], batch_size=BATCH_SIZE, shuffle=True, num_workers=2),
    'test': DataLoader(image_datasets['test'], batch_size=BATCH_SIZE, shuffle=False, num_workers=2)
}

dataset_sizes = {x: len(image_datasets[x]) for x in ['train', 'test']}
class_names = image_datasets['train'].classes

print(f"[INFO] Jumlah Data Training : {dataset_sizes['train']}")
print(f"[INFO] Jumlah Data Testing  : {dataset_sizes['test']}")
print(f"[INFO] Daftar Kelas          : {class_names}")
```

```python
# Load Model Inception-v4 Pretrained
model = timm.create_model('inception_v4', pretrained=True, num_classes=len(class_names))
model = model.to(device)

# Loss & Optimizer Adam (LR = 0.0001)
criterion = nn.CrossEntropyLoss()
optimizer = optim.Adam(model.parameters(), lr=0.0001)
NUM_EPOCHS = 3

# Fungsi Training
def train_model(model, criterion, optimizer, num_epochs=3):
    start_time = time.time()
    history = {'train_loss': [], 'train_acc': [], 'test_loss': [], 'test_acc': []}
    best_model_wts = copy.deepcopy(model.state_dict())
    best_acc = 0.0

    for epoch in range(num_epochs):
        print(f"\nEpoch {epoch + 1}/{num_epochs}")
        print('-' * 35)

        for phase in ['train', 'test']:
            if phase == 'train':
                model.train()
            else:
                model.eval()

            running_loss = 0.0
            running_corrects = 0

            for inputs, labels in dataloaders[phase]:
                inputs, labels = inputs.to(device), labels.to(device)
                optimizer.zero_grad()

                with torch.set_grad_enabled(phase == 'train'):
                    outputs = model(inputs)
                    _, preds = torch.max(outputs, 1)
                    loss = criterion(outputs, labels)

                    if phase == 'train':
                        loss.backward()
                        optimizer.step()

                running_loss += loss.item() * inputs.size(0)
                running_corrects += torch.sum(preds == labels.data)

            epoch_loss = running_loss / dataset_sizes[phase]
            epoch_acc = (running_corrects.double() / dataset_sizes[phase]) * 100.0

            history[f'{phase}_loss'].append(epoch_loss)
            history[f'{phase}_acc'].append(epoch_acc.item())

            print(f"{phase.capitalize()} Loss: {epoch_loss:.4f} | {phase.capitalize()} Acc: {epoch_acc:.2f}%")

            if phase == 'test' and epoch_acc > best_acc:
                best_acc = epoch_acc
                best_model_wts = copy.deepcopy(model.state_dict())

    print(f"\n[INFO] Selesai dalam {(time.time() - start_time) / 60:.2f} menit")
    print(f"[INFO] Testing Accuracy Tertinggi: {best_acc:.2f}%")

    model.load_state_dict(best_model_wts)
    return model, history

# Eksekusi Pelatihan
model, history = train_model(model, criterion, optimizer, num_epochs=NUM_EPOCHS)
```

```python
# Evaluasi Model
model.eval()
all_preds, all_labels = [], []

with torch.no_grad():
    for inputs, labels in dataloaders['test']:
        inputs, labels = inputs.to(device), labels.to(device)
        outputs = model(inputs)
        _, preds = torch.max(outputs, 1)
        all_preds.extend(preds.cpu().numpy())
        all_labels.extend(labels.cpu().numpy())

print("\n" + "="*50)
print("CLASSIFICATION REPORT")
print("="*50)
print(classification_report(all_labels, all_preds, target_names=class_names, digits=4))

# Visualisasi Confusion Matrix
cm = confusion_matrix(all_labels, all_preds)
plt.figure(figsize=(6, 5))
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', xticklabels=class_names, yticklabels=class_names)
plt.title('Confusion Matrix - Inception-v4')
plt.ylabel('Aktual')
plt.xlabel('Prediksi')
plt.tight_layout()
plt.savefig('confusion_matrix.png', dpi=300)
plt.show()

# Visualisasi Grafik Loss & Accuracy
epochs_range = range(1, NUM_EPOCHS + 1)
plt.figure(figsize=(12, 4))

plt.subplot(1, 2, 1)
plt.plot(epochs_range, history['train_acc'], 'o-', label='Train Acc')
plt.plot(epochs_range, history['test_acc'], 'o-', label='Test Acc')
plt.title('Grafik Accuracy Inception-v4')
plt.xlabel('Epoch')
plt.ylabel('Accuracy (%)')
plt.legend()
plt.grid(True)

plt.subplot(1, 2, 2)
plt.plot(epochs_range, history['train_loss'], 'o-', label='Train Loss')
plt.plot(epochs_range, history['test_loss'], 'o-', label='Test Loss')
plt.title('Grafik Loss Inception-v4')
plt.xlabel('Epoch')
plt.ylabel('Loss')
plt.legend()
plt.grid(True)

plt.tight_layout()
plt.savefig('grafik_kinerja.png', dpi=300)
plt.show()
```
